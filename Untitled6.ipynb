```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import mean_squared_error,r2_score
```

```python
diabetes=load_diabetes()
x=diabetes.data
y=diabetes.target
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
linear=LinearRegression()
linear.fit(x_train,y_train)
linear_pred=linear.predict(x_test)
linear_mse=mean_squared_error(y_test,linear_pred)
linear_r2=r2_score(y_test,linear_pred)
```

```python
ridge=Ridge()
ridge_parameters={"alpha":[0.1,1,10]}
ridge_cv=GridSearchCV(ridge,ridge_parameters,cv=5)
ridge_cv.fit(x_train,y_train)
ridge_model=ridge_cv.best_estimator_
ridge_pred=ridge_model.predict(x_test)
```

```python
lasso=Lasso(max_iter=10000)
lasso_parameters={"alpha":[0.1,1,10]}
lasso_cv=GridSearchCV(lasso,lasso_parameters,cv=5)
lasso_cv.fit(x_train,y_train)
lasso_model=lasso_cv.best_estimator_
lasso_pred=lasso_model.predict(x_test)
lasso_mse=mean_squared_error(y_test,lasso_pred)
lasso_r2=r2_score(y_test,lasso_pred)
```

```python
print("Best Ridge Alpha:",ridge_cv.best_params_["alpha"])
print("Best Lasso Alpha:",lasso_cv.best_params_["alpha"])
print("\nModel Performance:")
print("-"*55)
print(f"{'Model':<20}{'MSE':<15}{'R2 Score'}")
print("-"*55)
print(f"{'Linear Regression':<20}{linear_mse:<15.2f}{linear_r2:4f}")
print(f"{'Ridge Regression':<20}{ridge_mse:<15.2f}{ridge_r2:4f}")
print(f"{'Lasso Regression':<20}{lasso_mse:<15.2f}{lasso_r2:4f}")
```
